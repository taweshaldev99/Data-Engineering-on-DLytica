```python
import os

import getpass

import textwrap
 
import requests

from langchain_groq import ChatGroq
 
# Read the key from the environment if it's already set (e.g. exported in your shell,

# or set as a Colab/Kaggle secret). Otherwise, prompt for it without echoing it back.

# ChatGroq picks this up automatically from the environment — we never pass it in code.

if not os.environ.get("GROQ_API_KEY"):

    os.environ["GROQ_API_KEY"] = getpass.getpass("Enter your Groq API key: ")
 
print("GROQ_API_KEY is set.")
 
```

```python
# MODEL = "llama-3.3-70b-versatile"

MODEL="openai/gpt-oss-120b" # open-source model
 
 
# next word: probability prediction: [, ] -> high probability token

llm = ChatGroq(

    model=MODEL,

    temperature=0,   # 0 = deterministic/focused, 1+ = more random/creative

    max_tokens=2000,    # hard cap on how long the reply can be

)
 
response = llm.invoke("In two sentences, what makes Groq's inference fast?") 

print(response)

# print(response.content)

```
