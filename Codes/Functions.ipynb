# Parameters vs Arguments
- Parameter: the variable name listed in the function definition.
- Argument: the actual value you pass when calling the function.

```python
def add (a,b):
    result = a + b 
    print(f"{a} + {b} = {result}")

add(3,5)
add(8,9)
```

# Default Arguments
Default parameters allow functions to be called with fewer arguments. If an argument isn't provided, the
default is used:

```python
def mul(name, greeting ="Hello"):
    print(f"{greeting}, {name}!")

mul("Taweshal")
mul("Taweshal","Good evening")  #Overrides default
```
