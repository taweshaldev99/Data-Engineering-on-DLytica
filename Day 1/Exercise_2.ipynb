### Exercise 2: Data Quality Checker

```python
total_rows = 2000
missing_rows = 120
duplicate_rows = 30


Total_probelmatic_row = missing_rows + duplicate_rows
Percent_probelmatic_row = (Total_probelmatic_row / total_rows) * 100

print(f"Total rows: {total_rows}")
print(f"Total Problematic rowa: {Total_probelmatic_row}")
print(f"Total Problem Percentage : {Percent_probelmatic_row}")

if Percent_probelmatic_row <= 2 :
    print("Excellent")
elif Percent_probelmatic_row > 2 and Percent_probelmatic_row <= 5 :
    print("Acceptable")
else :
    print("Needs Cleaning")
```
