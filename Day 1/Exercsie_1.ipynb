```python
product_name = input("Write Product Name")
unit_price = int(input("What's the unit price ?"))
quantity_sold = int(input("What's the quantity sold ?"))
discount_percentage = float(input("Write Discount Percentage"))

print(f"Product: {product_name}")
print(f"unit_price: {unit_price}")
print(f"quantity_sold: {quantity_sold}")
print(f"discount_percentage: {discount_percentage}")

gross_sales = quantity_sold * unit_price
discount_amount = discount_percentage * gross_sales
Final_sales_amount = gross_sales - discount_amount

print(f"Product: {product_name}")
print(f"Gross Sales: {gross_sales}")
print(f"Discount: {discount_amount}")
print(f"Final Sales: {Final_sales_amount}")
```
