### Exercise 5: Pipeline Health Status

```python
rows_loaded = int(input("Enter rows_loaded: \n"))
rows_failed = int(input("Enter rows_failed: \n"))
runtime_minutes = int(input("Enter runtime_min: \n"))

total_row = rows_failed + rows_loaded

failure_rate = (rows_failed/ total_row ) * 100

if failure_rate <= 2 and runtime_minutes <=20 :
    print(f"Healthy, {failure_rate}%")
elif failure_rate <= 5 :
    print(f"Warning , {failure_rate}%")
else :
    print(f"Critical, {failure_rate}%")
```
