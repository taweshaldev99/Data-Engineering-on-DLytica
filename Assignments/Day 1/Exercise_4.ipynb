### Exercise 4: Customer Record Cleaner

```python
raw_name = "  sAgar THAPA "
raw_city = "kATHMANDU "
raw_age = "27"
raw_email = " SAGAR@MAIL.COM "

raw_name = raw_name.strip().title()
raw_city = raw_city.strip().title()
raw_email = raw_email.strip().lower()

raw_age = int(raw_age)

print(f"Name: {raw_name}")
print(f"raw_city: {raw_city}")
print(f"Age: {raw_age}")
print(f"raw_email: {raw_email}")

status = "Adult " if raw_age > 20 else "Young"
print(f"Status: {status}")

```
